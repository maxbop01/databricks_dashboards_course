# Lecture - Introduction to AI/BI Dashboards

## Overview

**AI/BI Dashboards** are Databricks' modern business intelligence experience. They let analysts turn governed Unity Catalog data into interactive, shareable reports that business users can explore on their own.

In this lecture you will get a short tour of the Databricks AI/BI family, learn what an AI/BI Dashboard actually is, and get familiar with the pieces you will use to build one. You will also see how **Genie Code**, Databricks' AI assistant, helps you author dashboards faster, and you will finish with a quick mental model for deciding **when to use a dashboard versus a Genie Agent**.

> This is an **overview lecture**. Publishing, scheduled refreshes, subscriptions, and embedding are covered in a later lecture.

## Learning Objectives

By the end of this lecture, you will be able to:

1. **Describe the components** in the Databricks AI/BI family (AI/BI Dashboards, Genie Agents, Genie) and how they fit together.
2. **Explain what an AI/BI Dashboard is** and identify its core building blocks (datasets, canvas, visualizations, filters, and pages).
3. **List the three data sources** a dashboard dataset can use: Unity Catalog tables, UC metric views, and dashboard local metric views.
4. **Describe Genie Code** and the AI-assisted authoring help it provides in the dashboard editor.
5. **Decide when to use a Dashboard vs. a Genie Agent** for a given analytics question.
6. **Identify the three ingredients** (purpose, audience, data) and **design best practices** for a well-scoped dashboard.

<details>

When you create a dataset, Databricks gives you a few options for where the data comes from. The right choice depends on whether the logic is **one-off** or **shared**.

#### Unity Catalog tables
The default. You query an existing UC table with SQL.

- Everything respects UC permissions, row filters, column masks, and lineage.
- Simple SQL is usually enough. Filter, join, aggregate in the dataset query.
- This is what we will use throughout this course.

#### Unity Catalog metric views
A **semantic layer** that lives in Unity Catalog alongside your tables.

- Centrally define business metrics (like revenue, gross margin, active users) along with their dimensions and formatting.
- Dashboards, Genie Agents, and alerts all read from the same metric view, so every tool reports the same number for "revenue".
- A dashboard can use a metric view the same way it uses a table.

#### Dashboard local metric views
Same semantic-layer idea, but **scoped to one dashboard**.

- Define measures, dimensions, and joins with a visual interface right in the dashboard editor.
- You do not need write access to Unity Catalog to create one.
- Great for an analyst who wants quick modeling without waiting for data engineering.
- Generally available. No Previews-portal enablement is required.

We will stick with UC tables for the hands-on demos in this course, and come back to dashboard local metric views in a later lecture.

</details>

##### EXPAND FOR ADDITIONAL NOTES

<details>

Think of Dashboards and Genie as two shapes of the same analytics experience.

#### Dashboards are for known, repeated questions

- You know who the audience is.
- You know the KPIs they want to see every week or month.
- You want the visual presentation to be consistent across time.
- The value is in the **curation**. An analyst decides what to show and how.

Typical examples: Canada Sales weekly review, Finance close-of-month deck, Product funnel tracker.

#### Genie is for unknown, one-off questions

- A business user looks at the dashboard and thinks "wait, why is that number down in Ontario?"
- That follow-up is probably not on any dashboard.
- In Genie they can just ask, and get a SQL-backed answer right away.
- The value is in the **accessibility**. Anyone can ask in plain English.

#### Most orgs use both

Dashboards are the anchor. They lay out the stable, agreed-upon view of the business. Genie is the exploration tool that answers the follow-ups dashboards cannot anticipate.

Both read the **same Unity Catalog data** and respect the same permissions, so the numbers agree.

</details>
