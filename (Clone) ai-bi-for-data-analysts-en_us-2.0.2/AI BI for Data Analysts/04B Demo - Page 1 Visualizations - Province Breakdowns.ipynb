# Demo B - Page 1 Visualizations - Province Breakdowns

## Overview

In the previous demo, **Demo A - Page 1 Visualizations - KPIs and Yearly Trends**, you built the top of Page 1 with a text header, three KPI counters, a combo chart, and an area chart. In this demo you'll complete Page 1 with a **section heading**, a **Sankey diagram** showing province-to-product flow, a **stacked bar chart** of sales by province, a **pivot table** for product-by-province detail, and a **corporate image** added from a Unity Catalog volume.

Along the way you'll use **Genie Code** to generate several of the visualizations from plain-English descriptions.

## Learning Objectives

By the end of this demo, you will be able to:

1. **Add a section-heading text widget** to split a dashboard page into visual zones.
2. **Use Genie Code** to create a Sankey diagram, a stacked bar chart, and a pivot table from natural-language descriptions.
3. **Fine-tune visualizations** in the configuration panel (titles, display names, Canadian currency, color scales, sort, legends).
4. **Embed an image** in a dashboard text widget using a Unity Catalog volume-stored asset.

### C2. Add a Sankey Chart - Province to Product by Order Amount

1. Return to the **Canada Sales Overview** page if you've navigated away.

2. Click the **visualization icon** in the palette.

3. Drop the new widget to the left below your section text box.

4. Set the widget to roughly **10H x 5W**.

5. Use **Genie Code** to create your Sankey widget!
    - Add the following to **Describe the visualization you want to create**:
        - `Create a sankey diagram by province and product name, using order amount as the value. Add a title.`

    Explore the **Configuration Panel**. You might have to make some minor modifications:

    | Setting | Value |
    |---------|-------|
    | **Title** | `Sales Flow: Province to Product` |
    | **Visualization** | Sankey |
    | **Stages** | - `FullOrdersData`.`province` <br>- `FullOrdersData`.`productname` |
    | **Value** | - `SUM(orderamt)` |

A <strong>Sankey diagram</strong> shows <strong>flow between categories</strong>. 

- On the left you see **provinces**

- On the right you see **products**

### C3. Add a Single Stacked Bar Chart - Sales by Province

1. Return to the **Canada Sales Overview** page if you've navigated away.

2. Click the **visualization icon** in the palette.

3. Place the new widget **to the right of the Sankey**, below the section text box.

4. Set the widget to **3H x remaining width**.

5. Use **Genie Code** to create your Stacked Bar widget.
    - Add the following to **Describe the visualization you want to create**:
        - `Create a stacked bar chart and use the sum of orderamt by province. Do not use any other columns. Make it a stacked bar chart. Add a title, labels and useful display names.`

6. **Genie Code** may not have returned exactly what we wanted (a **single grouped stacked bar**).

     In the **Configuration Panel** confirm or set the following:

    | Setting | Value |
    |---------|-------|
    | **Title** | `Province Contribution to Total Revenue` |
    | **Visualization** | Bar |
    | **X axis** | `FullOrdersData`.`SUM(orderamt)` (use the ↑ ↓ arrows on the axis field if `SUM(orderamt)` starts on the **Y axis**) |
    | **Y axis** | Empty (remove `province` if Genie placed it here) |
    | **Color** | `province` |
    | **Color Options** | - **Sort** > **By field** > **Ascending** <br>- **Legend Position**: Top <br>- **Show legend title**: `Province` |
    | **X axis options** | - **Display name**: `Total Sales` <br>- **Layout**: Stacked |
    | **Labels** | - **Stacked bars** - `Each segment` <br>- **Display** - `All` <br>- **Content** - `Field, SUM(orderamt)`

### C4. Add a Pivot Table - Product Name by Province

1. Return to the **Canada Sales Overview** page if you've navigated away.

2. Click the **visualization icon** in the palette.

3. Place the new widget **below the Stacked Bar**.

4. Size the widget to fill the space below the Stacked Bar, to the right of the Sankey.

5. Use **Genie Code** to create your Pivot Table widget.
    - Add the following to **Describe the visualization you want to create**:
        - `Create a pivot table. Use productname as the row, and province as the column, and the sum orderamt as the value. Add a color scale and format the values using Canadian dollars and compact.`

6. **Genie Code** may not have returned exactly what we wanted.

     In the **Configuration Panel** confirm or set the following:

    | Setting | Value |
    |---------|-------|
    | **Title** | `Sales by Product and Province` |
    | **Visualization** | Pivot |
    | **Rows** | `FullOrdersData`.`productname` |
    | **Columns** | `province` |
    | **Values** | - `SUM(orderamt)` <br>- Format the values. Select `SUM(orderamt)` <br>- **Style** > **Color ramp** > **Blues** <br>- **Format** > **Custom** > `$, Canadian dollars` <br>- **Format** > **Abbreviation** > `compact` |

### C5. Add a Corporate Image to the Page

In this step you'll add a corporate image (for example, a company logo or flag icon) next to the page header. The image is stored in a Unity Catalog volume: **labuser_UNIQUE_ID.aibi_dashboards.business_icon**.

Complete the following steps:

1. Navigate to your AI/BI Dashboard.

2. Reduce the **Canada Sales** header text widget two blocks to the left.

3. In the open space, add a **Text box**.

4. From your AI/BI Dashboard page, copy the Workspace URL from the beginning all the way after `.com` and paste it as the value for the `workspace_URL` SQL variable below.
   - **Example:** `https://databricks-instance.cloud.databricks.com`

5. Get the path to your image file using the workspace sidebar:
   - Select the **Catalog** icon
   - Expand the **labuser_UNIQUE_ID** catalog
   - Expand the **aibi_dashboards** schema
   - Expand **Volumes**
   - Expand **business_icon**
   - Right-click your image file and select **Copy volume file path**
   - Paste the value into the `img_vol_path` SQL variable below

6. Run the cell below and copy the output.
   - **Example output:** `https://my-workspace.cloud.databricks.com/ajax-api/2.0/fs/files/Volumes/labuser_UNIQUE_ID/aibi_dashboards/business_icon/<your-image>.png`

7. In your dashboard, add the image to the new text box:
   - Select the new blank text widget
   - In the toolbar, select the **Image** icon
   - Paste the copied image URL into the URL field
   - Select **Save**
   - Center align the image
   - **NOTE:** You can also download and upload the image.
