#### Dashboard relationships
- Dashboard relationships let you model the join logic between your datasets in AI/BI dashboards, so you can build multi-fact, multi-grain data models and reusable cross-dataset measures without pre-joining data in SQL. You model the relationship once, then use it across every visualization in the dashboard.
[AWS](https://docs.databricks.com/aws/en/dashboards/manage/data-modeling/dashboard-relationships/) |
[Azure](https://learn.microsoft.com/en-us/azure/databricks/dashboards/manage/data-modeling/dashboard-relationships/) |
[GCP](https://docs.databricks.com/gcp/en/dashboards/manage/data-modeling/dashboard-relationships/)

[GCP](https://docs.databricks.com/gcp/en/dashboards/parameters)
