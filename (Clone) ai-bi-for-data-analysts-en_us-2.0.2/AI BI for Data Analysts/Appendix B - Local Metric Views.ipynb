# Appendix B - Bonus Lab - Local Metric Views

## Overview

This bonus lab introduces **local metric views** in AI/BI Dashboards. A local metric view is a dashboard-scoped **semantic model** that lets you define **dimensions**, **measures**, and **joins** in a visual editor instead of hand-writing SQL for every visual.

Local metric views also carry **agentic metadata** (display names, formatting, comments, and synonyms) that make Genie and AI/BI smarter when generating charts and answering questions on your data.

Plan ~10 to 15 minutes for self-paced reading and exploration.

## Learning Objectives

By the end of this lab, you will be able to:

1. **Explain** what a local metric view is and how it differs from a SQL dataset and a Unity Catalog metric view.
2. **Identify** the benefits of defining a semantic model (dimensions, measures, joins) once and reusing it across visuals.
3. **Describe** agentic metadata (display names, formatting, comments, synonyms) and why it matters for Genie.
4. **Build** a local metric view from the Canada Sales course tables and use it in a visual.

## E. Build a Local Metric View

In this section you will recreate the **FullOrdersData** query you used in Notebook 03 as a **local metric view**, then use it in your dashboard.

The original SQL dataset looked like this:

```sql
SELECT
  c.customername,
  c.province,
  o.orderdate,
  YEAR(o.orderdate) AS year,
  p.productname,
  o.orderamt,
  o.quantity,
  o.salesrep
FROM ca_orders AS o
  INNER JOIN ca_customers AS c ON o.customerid = c.customerid
  INNER JOIN ca_products  AS p ON o.productid  = p.productid;
```


We will turn this into a local metric view with:
   -  **dimensions** (customer, province, date, product, sales rep)
   - **measures** (total sales, total quantity, orders count, average order value)
   -  **joins** between the three tables.
   - And add agentic metadata for the dashboard: **display names**, **formats**, **comments** and **synonyms**.
   



### E1. Create a Local Metric View for the Dashboard

1. **Create a new dashboard.** 
      - From the main navigation sidebar, right click **Dashboards** and select **Open Link in New Tab**.
      - Name the dashboard: `FirstName-LastInitials - Local Metric View`

2. Select the **Data** tab at the top of the dashboard.

3. **Add a dataset using a local metric view.**
   - Select **Add dataset** (not "Add SQL dataset"). 
   
4. In the **Select data source** picker, choose `labuser_UNIQUE_ID.aibi_dashboards.ca_orders` from Unity Catalog. 
      - This automatically creates a **local metric view** and opens it as a local metric view.
      - Name the **Dataset** `mvFullOrdersData`

5. Here you can write the metric view **YAML** or use the **UI**.

6. Select **UI** (Preview).
