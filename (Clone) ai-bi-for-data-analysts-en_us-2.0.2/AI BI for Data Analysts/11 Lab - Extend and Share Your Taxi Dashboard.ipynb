## Step 3. Polish What Genie Produced

1. Review what Genie Code generated and run each visual to confirm it works.

> **TIP:** If Genie picks the wrong chart type or aggregation, just ask it to refine. *"Make the daily chart a line, not a bar"* or *"sort the day of week chart Mon to Sun"* both work.

2. On your new page, look for improvements you can make like:

    - Rename any widget that still has a raw field name (`count(*)`, `monthly(date)`, etc.) to a business-friendly title.

    - Apply **number formatting** (no decimals on counts, currency on revenue, etc.) and **date formatting** where needed.

    - Resize and arrange widgets so the page reads cleanly.

    - Remove or modify widgets as you see fit.

    - Add another widget.

## Step 4. Add a Global Filter

A **global filter** applies to every page in the dashboard. Useful for things like a date range that should slice the whole story.

1. Add a **Filter** widget for `pickup_date` (date range).

2. Configure it to apply to **all available datasets**.
    - **NOTE:** Depending on the columns in the datasets, you might not be able to apply to all.

3. **Test it:** change the date range and confirm widgets on **both pages** update.
    - **NOTE:** The data is from **2016**. Use a date range within that year to see results.

## Step 5. Add a Page-Scoped Filter

A **page-scoped filter** only affects widgets on the page where it lives. Use it when a control is meaningful for one view but would clutter the others.

1. On the **Routes & Trip Duration** page, add a **Filter** widget for **dropoff_zip**.

2. Configure it to apply to **all possible available datasets**.
    - **NOTE:** Depending on the columns in the datasets, you might not be able to apply to all.

3. **Test it:** change the filter value and confirm the page updates. The **NYC Taxi Overview** page should not change.

## Step 7. Publish and Share

Now that you've finished editing, publish your dashboard!

> Note: this data is **not sensitive**.

1. Click **Publish** at the top of the dashboard.

    - Decide whether to publish with **embedded credentials** (viewers see the data through your access) or without (viewers need their own access).

2. Click **Share** and explore the options. Here you can decide who to share with and what access they have. For example:

    - A teammate, a workspace group, or yourself at a different email if you have one.

    - Give the principal **Can View** access **ONLY**.

3. In **Share** try the following:
    - Type `All workspace users`.
    - Then select `Can view`.
    - This would give all workspace users view access to your dashboard.
    - You can select **Cancel** for this training.

4. Open the **published** version (not in edit mode) in a new tab and confirm it renders as expected. Feel free to explore your published dashboard.
