# AI/BI for Data Analysts

## Overview

This course teaches data analysts how to design, build, publish, and operate AI/BI Dashboards in Databricks. AI/BI Dashboards combine governed Unity Catalog data with interactive visualizations, filters, and Genie integration so business users can explore answers without writing code.

The course follows a single end-to-end build. You start with source tables in Unity Catalog and finish with a published, monitored multi-page dashboard. Along the way you learn how dashboards fit into the broader Databricks AI/BI product family and where Genie, datasets, visualizations, and filters each fit in the workflow.

The content covers:
- AI/BI Dashboard fundamentals and how they relate to Genie and the rest of the Databricks platform.
- Exploring source data in Unity Catalog and designing reusable dashboard datasets with SQL.
- Authoring visualizations (KPIs, trends, breakdowns) and laying out a clean multi-page dashboard.
- Using Genie Code to draft SQL, charts, and filters from natural language prompts.
- Adding filters to make dashboards interactive and responsive to viewer questions.
- Publishing, sharing, and managing permissions so the right people can view and edit the dashboard.
- Running the dashboard in production with scheduled refresh, caching, and usage monitoring.

By the end of the course, you will be able to take a business question, design the supporting datasets, build a polished dashboard, and ship it to your audience with the correct permissions and a refresh strategy that keeps it accurate.

## Terminal Objectives

By the end of this course, you will be able to:

- **Explain** how AI/BI Dashboards fit into the Databricks AI/BI product family, and **distinguish** dashboards from Genie Agents and Genie Code in terms of audience and intended use.
- **Design** dashboard datasets on Unity Catalog tables using SQL.
- **Construct** a multi-page dashboard by **selecting** the appropriate visualization for each business question, including counters, combo charts, area charts, Sankey charts, and pivot tables.
- **Apply** Genie Code to generate visualizations and dataset SQL from natural language prompts, and **evaluate** the AI-generated output for accuracy, formatting, and presentation quality.
- **Configure** filters at the global, page, and widget scope, and **differentiate** when each scope is appropriate for an interactive dashboard.
- **Publish** a dashboard with the correct credential mode and sharing permissions, and **distinguish** between edit and published modes when iterating with stakeholders.
- **Operate** a dashboard in production by **configuring** refresh schedules, **interpreting** caching behavior, and **applying** a production-readiness checklist before audience handoff.

## A. Prerequisites

Before starting this course, learners should be comfortable with the following:

- **Basic familiarity with the Databricks workspace.** You can navigate the workspace UI, open and use notebooks, and attach a SQL warehouse or other compute resource.

- **Basic familiarity with Unity Catalog.** You understand the three-level namespace (catalog, schema, table) and can find a table in the catalog explorer.

- **Working knowledge of SQL, including joins.** You can read and write SELECT statements with WHERE, GROUP BY, and aggregates, and you can join two tables on a shared key.
