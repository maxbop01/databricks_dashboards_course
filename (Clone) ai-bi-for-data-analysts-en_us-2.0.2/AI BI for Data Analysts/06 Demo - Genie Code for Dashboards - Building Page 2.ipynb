# Demo - Genie Code for Dashboards - Building Page 2

## Overview

In the previous demos you built **Page 1** widget by widget, with only a few Genie Code assists. This demo flips that model entirely. You'll **prompt Genie Code once at the page level** and watch it plan, generate, and lay out **every widget on Page 2 in one pass**.

You'll use the **Opportunities** dataset (sales pipeline data), including the **deal_size** custom calculation you added in an earlier demonstration. The goal is not to build four widgets separately. The goal is to **describe the page you want, and let Genie build it**.

What you'll see:

1. The difference between a **weak page prompt** and a **specific page prompt**.
2. How to structure a multi-widget prompt so Genie produces a coherent layout.
3. How to clean up Genie's output in the configuration panel rather than re-prompting.

## Learning Objectives

By the end of this demo, you will be able to:

1. **Add a new page** to an existing AI/BI Dashboard.
2. **Open Genie Code at the page scope** (not tied to a single widget).
3. **Write a structured page-level prompt** that specifies every widget plus layout.
4. **Review and clean up** a Genie-generated page in the configuration panel.
5. **Reuse a custom calculation** (`deal_size` from Demo 03) in Genie-generated charts.

## E. Developer Review and Clean Up




### E1. Developer Review

Genie is fast but not perfect. Expect to spend time reviewing and polishing the result.

##### 5 things to look for after Genie generates a page:

1. **Titles and labels.** Widget titles, axis labels, and legend entries often show raw field names like `sum(amount)` or `monthly(date)`. Rename them to business-friendly labels.

2. **Number formatting.** Currency symbols, decimal precision, percentages, and units rarely match your audience by default. Set the right format on every measure.

3. **Sort order.** Bar charts default to alphabetical, donut slices to load order. Re-sort to whatever tells the story (descending value, manual category order, etc.).

4. **Conditional styling.** Counters and table cells usually need color rules (green for good, red for bad, etc.) to convey status at a glance. Genie rarely adds these.

5. **Chart type and layout.** Grouped vs stacked bars, line vs area trends, donut vs pie. Genie picks something reasonable, not always the best choice for the question being answered.


### E2. Developer Updates
After Genie builds your dashboard, it's up to you to review and update. Explore your dashboard and find 2-3 items to clean up for your production dashboard


1. Stay on the page Genie created.

2. Apply a few general fixes and updates that you would like. Typical updates to try:
   - Rearrange the widgets so the most important ones are top-left.
   - Update widget display options, titles, display names, values, and descriptions.
   - Add or remove widgets so the page tells a focused story.
      - Try adding a heatmap of `Total Sales Heatmap by Province and Deal Phase`.
   - Apply Canadian currency formatting and sort orders where missing.

3. After a few minor modifications, your finished page could look roughly like the checkpoint image below.
