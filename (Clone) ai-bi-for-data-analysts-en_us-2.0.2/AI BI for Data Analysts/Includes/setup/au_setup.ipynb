```python
%sql
USE CATALOG dbacademy;  -- Modify to use DBACADEMY in Vocareum
USE SCHEMA ops;

-- Creates View of keys + values
CREATE OR REPLACE TEMP VIEW user_info AS
SELECT map_from_arrays(collect_list(key), collect_list(value))
FROM meta;

-- Creates MAP object of ops table key + values using the view
DECLARE OR REPLACE DA MAP<STRING,STRING>;
SET VAR DA = (SELECT * FROM user_info);

-- Set default schema for the user  
USE SCHEMA IDENTIFIER(DA.schema_name);

-- Copy in tables from their masters in Marketplace

CREATE OR REPLACE TABLE au_opportunities AS 
  SELECT * FROM dbacademy_databricks_simulated_australia_sales_and_opportunities_data.v01.opportunities;
CREATE OR REPLACE TABLE au_orders AS 
  SELECT * FROM dbacademy_databricks_simulated_australia_sales_and_opportunities_data.v01.orders;
CREATE OR REPLACE TABLE au_customers AS 
  SELECT * FROM dbacademy_databricks_simulated_australia_sales_and_opportunities_data.v01.customers;

-- Show the user their catalog + schema, since they will need to interpolate their schema into lab instructions
SELECT current_catalog() as CourseCatalog, current_schema() as YourSchema;
```
