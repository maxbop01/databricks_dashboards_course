```python
from pathlib import Path

## Gets the path to the current folder
current_path = Path.cwd()

## Reads the Solution Demonstration Dashboard - Canadian Sales.lvdash.json and adds your catalog + schema 
updated_json_template = load_and_rewrite_template(
    template_path = f'{current_path}/TEMPLATE Solution Demonstration Dashboard - Canadian Sales.lvdash.json', 
    ORIGINAL_CATALOG = ORIGINAL_CATALOG,
    ORIGINAL_SCHEMA =  ORIGINAL_SCHEMA,
    catalog = my_catalog, 
    schema= ORIGINAL_SCHEMA
)

## Obtains your Warehouse ID based of warehouse name (If using outside of the provided labs you will have to modify the name of your shared warehouse)
warehouse_id = find_warehouse_id_by_name(warehouse_name='shared_warehouse')

## Sets your solution dashboard name "labuser_YOUR_UNIQUE_ID"
dashboard_name = f'{my_catalog} - Solution - Canadian Sales'

## Uses the SDK to import the updated .lvdash.json file.
create_solution_dashboard(
    catalog = my_catalog,
    schema = 'aibi_dashboards',
    display_name = dashboard_name,
    parent_path = str(current_path),
    serialized_json = updated_json_template,
    warehouse_id = warehouse_id
)

## Displays config variables
display_config_values(
    [
        ("Your Catalog", my_catalog),
        ("Your Schema", "aibi_dashboards"),
        ("Your Demonstration Solution Dashboard Name", dashboard_name)
    ]
)
```
