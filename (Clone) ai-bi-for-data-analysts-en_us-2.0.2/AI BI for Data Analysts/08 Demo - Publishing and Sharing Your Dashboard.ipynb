# Demo - Publishing and Sharing Your Dashboard

## Overview

A dashboard that only you can see is a draft. This demo walks through the last mile: explore the **Settings** panel, **publish** the dashboard, **certify** it as trusted, **share** it with the right people at the right level of access, view it as a consumer, and point you to two extensions (**Genie Agents** and **embedded dashboards**).

## Learning Objectives

By the end of this demo, you will be able to:

1. Explore the dashboard's **Settings** panel and apply a theme.
2. **Publish** a dashboard and toggle between **Draft** and **Published** views.
3. Choose between **Share data permissions** (embedded credentials) and **Individual data permissions**.
4. **Certify** a dashboard so consumers know it is trusted.
5. Set **sharing permissions** (Can View / Can Run / Can Edit / Can Manage) for the right audience.
6. View the dashboard from the **consumer experience** in both the workspace and **Genie**.
7. Identify when to extend with a **Genie Agent** or **embed** the dashboard.

function selectComp(id) {
  var wrap = document.getElementById('detail-wrap');
  var card = document.getElementById('detail-card');
  var c = COMPONENTS[id];

  document.querySelectorAll('.perm-box').forEach(function(b) {
    var isActive = parseInt(b.dataset.id, 10) === id;
    b.classList.toggle('active', isActive);
  });

  if (current === id) {
    wrap.classList.remove('open');
    document.querySelectorAll('.perm-box').forEach(function(b) {
      b.classList.remove('active');
    });
    current = null;
    return;
  }

  current = id;

  card.style.borderTopColor = c.color;
  card.innerHTML =
    '<div style="font-size:18pt;font-weight:700;margin-bottom:14px;color:#0b2026;">' + c.title + '</div>'
    + c.content
    + '<div class="example-block" style="background:' + c.exampleBg + ';border-color:' + c.color + ';">'
    + '<strong>Example</strong>' + c.example
    + '</div>';

  wrap.classList.add('open');

### C4. Toggle Between Edit and Published Mode

After publishing, a view toggle appears near the dashboard title. The label depends on which mode you are currently in:

   - **View Published** appears when you are in **Edit** mode (you should see this now, since you just published from Edit mode).

   - **Edit Draft** appears when you are in **Published** mode.

2. Notice this is what consumers see. You cannot edit widgets or pages in **Published** mode.

3. Click **Edit Draft** to return to authoring.


### C5. Modify a Published Dashboard and Republish

### E1. Choose Your Audience (Share)

The **Share** button gives you three audience scopes:
  - **All workspace users**, everyone with access to this workspace (`All workspace users`).
  - **Anyone in my account can view**, dashboard-only access for account users who don't have full workspace access (*great for business consumers*).
  - **Named individuals or groups**, the tightest option, use for sensitive dashboards.

1. From the **Share** dialog (it opens automatically after publish, or click **Share** next to the **Publish** button), pick a permission level and an audience scope.

2. Click **Save** and close the **Share** dialog.

> **Reminder:** Dashboard permissions are separate from Unity Catalog data permissions. Under **Share data permissions**, viewers use the publisher's UC credentials. Under **Individual data permissions**, each viewer must have their own UC access to the underlying tables.

## G. Genie Agents

Treat the dashboard as the curated view, the Genie Agent as the open-ended explorer.

1. Go into your **Published** dashboard.

2. Find the **Ask Genie** button.
