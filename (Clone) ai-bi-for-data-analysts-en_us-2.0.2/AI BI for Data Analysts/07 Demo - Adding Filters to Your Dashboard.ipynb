# Demo - Adding Filters to Your Dashboard

## Overview

Filters let viewers explore a dashboard without you having to build new widgets. One person finds "2024 in Ontario," another finds "closed-won deals," both from the same dashboard. In this demo you'll add three filters to the Canada Sales dashboard and see how the **scope** of each one changes what viewers can do.

## Learning Objectives

By the end of this demo, you will be able to:

1. Describe the **three filter scopes**: global, page-level, and widget-level.
2. Add a **global filter** that drives widgets across both pages of the dashboard.
3. Add **page-level filters** that drive only one page each.
4. Use **cross-filtering** to explore a chart by clicking its elements.
5. Add a **widget-level filter** that pins a static view to one widget.

## C. Add a Global Filter: Year

A **Year** filter lets viewers focus the dashboard on one calendar year at a time. Since Year applies to widgets on both pages, it belongs in the **global** scope.

1. Navigate to your dashboard. Select the **Canada Sales Overview** page.

2. Click the **filter icon** in the upper-left corner of the dashboard canvas to open the **global filter panel**.

3. Click the **+** button in the panel. The filter configuration panel opens on the right.

4. Configure the filter:

    | Setting | Value |
    |---------|-------|
    | **Title** | Checked, rename to `Year` |

## F. Try Cross-filtering

Cross-filtering lets viewers filter a page by **clicking a chart element** instead of the filter widgets. It works on any bar, pie, heatmap, box plot, histogram, scatter, or point map that uses the same dataset.

1. Stay on the **Opportunities Pipeline** page. Clear any set filters.

2. Click on a single **phase segment** like `1-Active` in one of the widgets:
    - If you have the **Pipeline Funnel by Phase** chart select `1-Active`.

3. Watch all the other Page 2 widgets (pie, area, bar charts, detail table) update immediately to show only that phase.

4. A **cross-filter chip** appears in the active filter bar at the top of the page. Click the X on the chip to clear the cross-filter.

    - **NOTE:** Cross-filtering stacks with your existing Phase filter. If you already have `1-Active` selected in the Phase filter and then click `1-Active` again in the funnel, nothing changes. Click a different phase to see the interaction.
