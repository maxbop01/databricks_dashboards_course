# Demo A - Page 1 Visualizations - KPIs and Yearly Trends

## Overview

Databricks AI/BI Dashboards help you quickly transform data into shareable insights. In the previous demo you built the **datasets** that feed the dashboard. In this demo (the first of two parts for Page 1), you'll populate the top of the **Canada Sales Overview** page with text, KPI counters, and year-level charts.

You'll first explore the main **visualization types** available in AI/BI, then add a page header, a **row of three KPI counters** (Total Sales vs Target, Total Orders, Units Sold), a **combo chart** of yearly sales vs target, and an **area chart** of monthly sales by product.

The remaining widgets (section heading, Sankey, stacked bar, pivot, corporate image) live in **Part 2**.

## Learning Objectives

By the end of this demo, you will be able to:

1. **Recognize the major visualization types** available in AI/BI Dashboards.
2. **Add a text widget** to the dashboard canvas and format it with Markdown.
3. **Add a comparison counter widget** with a target value, currency formatting, and conditional styling.
4. **Add display-only counter widgets** for headline KPIs.
5. **Build a combo chart** that overlays a line (target) on bars (actual).
6. **Use Genie Code** to generate an area chart from a natural-language description.

## D. Build the `Canada Sales Overview` Page

1. Confirm you are in your **Canada Sales Overview** page.

- **NOTES:**

    - The text box contains a variety of option you can use to format your text.

### D2. Add a Comparison Counter Widget - Total Sales by Sales Target

You'll display **current sales** against the **sales target** for the latest year in the data.

1. In the bottom palette, click the **Add a visualization** icon.

2. Place the widget at the **left side of the canvas, below the text box**, sized **3H x 4W**. Click it to give it focus.

3. In the **Configuration Panel** on the right, set:

    | Setting | Value |
    |---------|-------|
    | **Title** | Checked |
    | **Description** | Checked |
    | **Visualization** | Counter |
    | **Value** | - Select `SalesTargets`.`actual_sales` <br>- Click the field and change `SUM(actual_sales)` to **None**, since the `SalesTargets` query already aggregated this column. <br>- **Row number** - `1` (our query sorted by year descending, so row 1 is the latest year) |
    | **Comparison** | - Select `sales_target` <br>- **Row number** - `1` (target for the latest year) <br>- **Change** - `Number` |


4. Select the **Data** tab and run the `SalesTargets` query to **confirm the counter numbers match row 1**.

5. Format the **actual_sales** value as Canadian currency:
    - Click `actual_sales` in the **Value** area.
    - Select **Format** > **Custom**.
    - Set **Type** to **Currency ($)**, then choose **Canadian Dollars ($)** from the dropdown.
    - Leave the format panel open for the next step.

6. Add a **conditional style** so the counter turns green when on track and red when below target:
    - Select **Style**.
    - Click the **+** next to **Conditional Style**.
    - Set **If Value >= Target** to **Green**.
    - Add another rule: **If Value < Target** to **Red**.

7. Click the **Widget Title** and rename it to:
    -  `Current Total Sales by Sales Target`.

7. Click the **Description** and set it to:
    -  `Defaults to the latest year. Updates when a year filter is applied.`

8. Format the **sales_target** comparison value the same way as step 5: **Currency ($)**, **Canadian Dollars ($)**.

  <strong style="display:block; color:#0d47a1; margin-bottom:6px; font-size: 1.1em;">
    Why a comparison counter?
  </strong>
  <div style="color:#333;">

A <strong>counter</strong> is the simplest, highest-signal visualization: one big number. 

Adding a <strong>comparison</strong> value makes the number meaningful so viewers can instantly see whether you're ahead of or behind target without reading a chart.

<strong>About "Row number":</strong>
- Our <code>SalesTargets</code> dataset has one row per year, sorted by year descending

- That means row 1 is the latest year. Setting <strong>Row number = 1</strong> on both Value and Comparison picks the current-year numbers without writing a <code>WHERE</code> clause. When you add a Year filter in a later demo, this counter will update automatically.

<strong>About conditional style:</strong>

- The green/red rules tie the visual color to the story. Green at a glance = on track, red = off track. Viewers don't have to do the math.

  </div>
</div>

### D3. Add Two More Counter Widgets

You currently have **one** counter, the **Total Sales by Sales Target**. 

Let's add **two more** so the top of the dashboard shows **three counters across a single row**:

| # | Counter | What it shows |
|---|---------|---------------|
| 1 | COMPLETE - **Total Sales by Sales Target** | Latest-year revenue vs. target (from **D2**) |
| 2 | **Total Orders** | Total number of orders placed (all years) |
| 3 | **Total Units Sold** | Total units shipped (all years) |

The two new counters share the **FullOrdersData** dataset you created earlier.

For **each** of the counters below, repeat this pattern:

1. Click the **visualization icon** in the palette.

2. Drop the widget to the **right** of the previous counter, so all three sit across one row at the top of the canvas.

3. Click the widget to give it focus and configure it using the row in the table below.

4. Click **Widget Title** and rename it to the **Title** in the table.

5. Set the following for each **Counter** widget.

| Counter # | Title | Visualization | Value | Format |
|---|---------|---------------|-------|--------|
| 2 | **Total Orders** | Counter | `FullOrdersData`.`COUNT(*)` | **Abbreviation**: `NONE` |
| 3 | **Total Units Sold** |Counter | `FullOrdersData`.`SUM(quantity)` | **Abbreviation**: `Compact` |

6. After all three counters are placed, drag their edges so they are the **same height** and **evenly sized** across the row.

**NOTE:** Unlike the previous counter, we decided that these two do **not** get a **Comparison** value. **These are display-only KPIs.**
